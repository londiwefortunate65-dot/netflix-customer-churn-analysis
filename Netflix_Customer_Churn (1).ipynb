```python
df['login_activity_group'] = pd.cut(
    df['last_login_days'],
    bins=[-1, 7, 14, 30, 60, float('inf')],
    labels=[
        '0-7 days',
        '8-14 days',
        '15-30 days',
        '31-60 days',
        'Over 60 days'
    ]
)

login_risk = df.groupby(
    'login_activity_group', observed=False
).agg(
    customers=('customer_id', 'nunique'),
    churn_rate=('churned', 'mean')
)

login_risk['churn_rate'] = (
    login_risk['churn_rate'] * 100
).round(2)

login_risk
```

```python
tier_analysis = df.groupby('subscription_type').agg(
    customers=('customer_id', 'nunique'),
    churned_customers=('churned', 'sum'),
    churn_rate=('churned', 'mean'),
    average_monthly_fee=('monthly_fee', 'mean')
)

tier_analysis['churn_rate'] = (
    tier_analysis['churn_rate'] * 100
).round(2)

tier_analysis.round(2)
```

```python

import matplotlib.pyplot as plt
import numpy as np

# Create the summary table
customer_profile = df.groupby('churned').agg(
    total_customers=('customer_id', 'nunique'),
    average_age=('age', 'mean'),
    minimum_age=('age', 'min'),
    maximum_age=('age', 'max')
).round(2)

# Rename customer status
customer_profile.index = customer_profile.index.map({
    1: 'Churned',
    0: 'Retained'
})

# Create separate charts for each metric
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

metrics = [
    ('total_customers', 'Total Customers', 'Count'),
    ('average_age', 'Average Age', 'Years'),
    ('minimum_age', 'Minimum Age', 'Years'),
    ('maximum_age', 'Maximum Age', 'Years')
]

colors = ['mediumpurple', 'rebeccapurple']

for ax, (column, title, ylabel) in zip(axes.flat, metrics):
    values = customer_profile[column]

    bars = ax.bar(
        values.index,
        values.values,
        color=colors,
        edgecolor='white'
    )

    ax.bar_label(bars, fmt='%.2f', padding=3, fontsize=9)
    ax.set_title(title, fontweight='bold')
    ax.set_ylabel(ylabel)
    ax.set_xlabel('Customer Status')

    # Add space above the labels
    ax.set_ylim(0, values.max() * 1.15)

fig.suptitle(
    'Netflix Customer Profile: Churned vs Retained',
    fontsize=15,
    fontweight='bold'
)

plt.tight_layout()

plt.savefig("netflix_visuals/retained vs churn.png",
            dpi=300,
            bbox_inches="tight")
plt.show()



```

```python

import matplotlib.pyplot as plt
import numpy as np

# Exclude categories with zero customers or missing churn rates
plot_data = login_risk[
    (login_risk['customers'] > 0) &
    (login_risk['churn_rate'].notna())
]

fig, ax1 = plt.subplots(figsize=(11, 6))

x = np.arange(len(plot_data))

# Bar chart: customer counts
bars = ax1.bar(
    x,
    plot_data['customers'],
    color='turquoise',
    width=0.65,
    label='Customers'
)

ax1.set_xlabel('Days Since Last Login', fontsize=11, labelpad=12)
ax1.set_ylabel('Number of Customers', fontsize=11)
ax1.set_xticks(x)
ax1.set_xticklabels(plot_data.index.astype(str))
ax1.bar_label(bars, fmt='{:,.0f}', padding=3, fontsize=9)

# Line chart: churn rate on a second axis
ax2 = ax1.twinx()

ax2.plot(
    x,
    plot_data['churn_rate'],
    color='teal',
    marker='o',
    linewidth=2.5,
    markersize=7,
    label='Churn Rate (%)'
)

# Add percentage labels
for i, value in enumerate(plot_data['churn_rate']):
    ax2.annotate(
        f'{value:.2f}%',
        (x[i], value),
        xytext=(0, 10),
        textcoords='offset points',
        ha='center',
        fontsize=10,
        fontweight='bold',
        color='indigo'
    )

ax2.set_ylabel('Churn Rate (%)', fontsize=11)
ax2.set_ylim(0, 100)

# Title and combined legend
plt.title(
    'Netflix Customer Churn Rate by Login Activity',
    fontsize=14,
    fontweight='bold',
    pad=18
)

handles1, labels1 = ax1.get_legend_handles_labels()
handles2, labels2 = ax2.get_legend_handles_labels()

ax1.legend(
    handles1 + handles2,
    labels1 + labels2,
    loc='upper left'
)

plt.tight_layout()

plt.savefig("netflix_visuals/login_activity.png",
            dpi=300,
            bbox_inches="tight")

plt.show()
```

```python

import pandas as pd

# 1. Overall customer figures
total_customers = df['customer_id'].nunique()
churned_customers = df.loc[
    df['churned'] == 1, 'customer_id'
].nunique()
retained_customers = df.loc[
    df['churned'] == 0, 'customer_id'
].nunique()

overall_churn_rate = df['churned'].mean() * 100
average_age = df['age'].mean()

# 2. Subscription analysis
subscription_summary = df.groupby('subscription_type').agg(
    customers=('customer_id', 'nunique'),
    churn_rate=('churned', 'mean'),
    average_monthly_fee=('monthly_fee', 'mean')
)

subscription_summary['churn_rate'] *= 100
subscription_summary = subscription_summary.round(2)

# 3. Age-group analysis
age_summary = df.groupby('age_group', observed=True).agg(
    customers=('customer_id', 'nunique'),
    churn_rate=('churned', 'mean')
)

age_summary['churn_rate'] = (
    age_summary['churn_rate'] * 100
).round(2)

# 4. Profile analysis by subscription type
profile_analysis = df.groupby(
    ['subscription_type', 'number_of_profiles']
).agg(
    customers=('customer_id', 'nunique'),
    churn_rate=('churned', 'mean')
)

profile_analysis['churn_rate'] = (
    profile_analysis['churn_rate'] * 100
).round(2)

# 5. Device analysis
device_analysis = df.groupby('device').agg(
    customers=('customer_id', 'nunique'),
    churn_rate=('churned', 'mean')
)

device_analysis['churn_rate'] = (
    device_analysis['churn_rate'] * 100
).round(2)

device_analysis = device_analysis.sort_values(
    'churn_rate', ascending=False
)

# 6. Payment method analysis
payment_analysis = df.groupby('payment_method').agg(
    customers=('customer_id', 'nunique'),
    churn_rate=('churned', 'mean')
)

payment_analysis['churn_rate'] = (
    payment_analysis['churn_rate'] * 100
).round(2)

payment_analysis = payment_analysis.sort_values(
    'churn_rate', ascending=False
)

# 7. Engagement comparison
engagement_summary = df.groupby('churned').agg(
    customers=('customer_id', 'nunique'),
    average_watch_hours=('watch_hours', 'mean'),
    average_days_since_login=('last_login_days', 'mean'),
    average_daily_watch_time=('avg_watch_time_per_day', 'mean'),
    average_profiles=('number_of_profiles', 'mean')
).round(2)

engagement_summary.index = engagement_summary.index.map(
    {0: 'Retained', 1: 'Churned'}
)

# Display all results
print("OVERALL RESULTS")
print("Total customers:", total_customers)
print("Churned customers:", churned_customers)
print("Retained customers:", retained_customers)
print("Overall churn rate:", round(overall_churn_rate, 2), "%")
print("Average customer age:", round(average_age, 2))

print("\nSUBSCRIPTION SUMMARY")
display(subscription_summary)

print("\nAGE-GROUP SUMMARY")
display(age_summary)

print("\nPROFILE ANALYSIS")
display(profile_analysis)

print("\nDEVICE ANALYSIS")
display(device_analysis)

print("\nPAYMENT METHOD ANALYSIS")
display(payment_analysis)

print("\nENGAGEMENT SUMMARY")
display(engagement_summary)
```


# Netflix Customer Churn Analysis

## 1. Business Summary

### Project Objective
The objective of this project was to analyse customer churn and identify factors associated with customer retention using Python and Pandas.

### Key Findings
- **Overall retention:** Compare the total numbers of retained and churned customers.
- **Subscription type:** Basic subscribers had a churn rate of 61.83%, Standard 45.44%, and Premium 43.71%.
- **Age groups:** Customers aged 35–44 had the lowest churn rate at 48.56%, while ages 45–54 had the highest at 53.97%.
- **Customer engagement:** Lower viewing activity and longer periods since the last login were associated with higher churn.
- **Profile usage:** Higher profile counts appeared to be associated with better retention, particularly among Premium subscribers.
- **Device usage:** TV and desktop users had lower observed churn than users of some portable devices.
- **Payment methods:** PayPal, gift cards, and cryptocurrency showed higher churn in the analysis and warrant further investigation.
- **Average age:** The dataset's average customer age was approximately 43 years.

## 2. Business Recommendations

1. Investigate why Basic subscribers have higher churn than Premium subscribers.
2. Test re-engagement campaigns for customers with declining viewing activity or prolonged login inactivity.
3. Compare engagement patterns across age groups to identify opportunities for improving retention.
4. Investigate device-related differences and possible technical or usability issues.
5. Examine payment-method patterns without assuming affordability is the cause.
6. Investigate the relationship between profile usage and retention.
7. Monitor churn and retention metrics regularly to evaluate whether interventions work.

## 3. Conclusion

The analysis identified subscription type, customer engagement, age group, profile usage, device, and payment method as areas for further investigation. Retention strategies should be tested and evaluated using measurable outcomes.

The findings describe associations in this dataset and do not prove that any individual factor directly causes customer churn.
